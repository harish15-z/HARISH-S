```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. ஒரு சாம்பிள் டேட்டாவை உருவாக்குவோம் (குப்பையான டேட்டாவுடன்)
data = {
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Tablet', 'Mobile', None, 'Tablet'],
    'Sales': [50000, 20000, 50000, 15000, None, 10000, 15000],
    'Date': ['2026-01-01', '2026-01-02', '2026-01-01', '2026-01-03', '2026-01-04', '2026-01-05', '2026-01-03']
}

df = pd.DataFrame(data)
print("--- கிளீனிங் செய்வதற்கு முன் ---")
print(df)
print("\n")

# 2. Data Cleaning - டூப்ளிகேட்களை நீக்குதல்
df = df.drop_duplicates()

# 3. Data Cleaning - காலியாக உள்ள (None) இடங்களை நிரப்புதல்
df['Product'] = df['Product'].fillna('Unknown')
df['Sales'] = df['Sales'].fillna(df['Sales'].mean()) # விடுபட்ட விற்பனைக்கு ஆவரேஜ் போடுதல்

print("--- கிளீனிங் செய்த பின் ---")
print(df)

# 4. Visualization - ஒரு அழகான சார்ட்
plt.figure(figsize=(8,5))
sns.barplot(x='Product', y='Sales', data=df, palette='viridis')
plt.title('Product vs Sales (Cleaned Data)')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load the dataset
# Replace 'your_file_name.csv' with your actual file name
file_path = 'sample_data/california_housing_test.csv'

try:
    df = pd.read_csv(file_path)

    # 2. View basic info
    print("Dataset Preview:")
    display(df.head())

    # 3. Check for Missing Values
    print("\nMissing Values Count:")
    print(df.isnull().sum())

    # 4. Data Cleaning (Filling missing values with 0)
    df_cleaned = df.fillna(0)

    # 5. Visualization - Outliers Detection
    plt.figure(figsize=(12,6))
    sns.boxplot(data=df_cleaned.select_dtypes(include=['number']))
    plt.title('Outlier Analysis - Cleaned Data')
    plt.xticks(rotation=45)
    plt.show()

    # 6. Basic Statistics
    print("\nStatistical Summary:")
    display(df_cleaned.describe())

except Exception as e:
    print(f"Error occurred: {e}")
```

```python
# 1. Correlation Heatmap - To see how columns relate to each other
plt.figure(figsize=(10, 8))
correlation_matrix = df_cleaned.select_dtypes(include=['number']).corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Features')
plt.show()

# 2. Distribution Plot - To see the spread of a specific column
# Replace 'median_income' with a column name from your dataset
target_column = df_cleaned.select_dtypes(include=['number']).columns[0]
plt.figure(figsize=(8, 5))
sns.histplot(df_cleaned[target_column], kde=True, color='blue')
plt.title(f'Distribution of {target_column}')
plt.show()

# 3. Saving the Cleaned Data
df_cleaned.to_csv('cleaned_project_data.csv', index=False)
print("Success! Your cleaned data is saved as 'cleaned_project_data.csv'")
```
